# Math for learning

A vector is an ordered list of numbers. A dot product multiplies matching entries and sums them; weighted sums are the basis of many models. A derivative measures local change, and a gradient collects derivatives. A probability describes uncertainty on a 0–1 scale; it is not a guarantee.

Run the cells in order. Change values and rerun to explore.

```python
x = [2, 3]
w = [0.5, -1]
print("dot product:", sum(a*b for a,b in zip(x,w)))
h = 0.0001
print("slope of x² at 3:", ((3+h)**2-3**2)/h)
```
