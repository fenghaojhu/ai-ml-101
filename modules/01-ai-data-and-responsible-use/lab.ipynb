# AI, data and responsible use

A model learns a mapping from examples. A feature is information available at prediction time; a label is the outcome to predict. Training examples approximate a population, so collecting only convenient examples can leave groups out. A model can be accurate overall and still harm people in a subgroup.

Run the cells in order. Change values and rerun to explore.

```python
data = [("sunny", 0), ("rainy", 1), ("rainy", 1), ("sunny", 0)]
for weather in ("sunny", "rainy"):
    labels = [late for w, late in data if w == weather]
    print(weather, "delay rate:", sum(labels) / len(labels))
```
