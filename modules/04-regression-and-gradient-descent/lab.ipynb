# Regression and gradient descent

Linear regression predicts y = wx + b. Mean squared error averages squared prediction errors. Gradient descent adjusts weights in the direction that reduces loss locally; the learning rate controls step size. A low training loss alone does not establish generalization.

Run the cells in order. Change values and rerun to explore.

```python
xs, ys = [0., 1., 2., 3.], [1., 3., 5., 7.]
w = b = 0.0
for step in range(300):
    errors = [w*x+b-y for x,y in zip(xs,ys)]
    dw = sum(2*e*x for e,x in zip(errors,xs))/len(xs)
    db = sum(2*e for e in errors)/len(xs)
    w -= 0.05*dw; b -= 0.05*db
print("w, b:", round(w,3), round(b,3), "prediction at 4:", round(4*w+b,3))
```
